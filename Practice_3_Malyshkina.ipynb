# Выживаемость пассажиров внутри класса обслуживания

Доля выживших пассажиров среди всех пассажиров данного класса обслуживания:

```python
base_set.groupby('Pclass')['Survived'].mean() * 100
```

Доля пассажиров каждого класса среди всех выживших:

```python
base_set.groupby('Pclass')['Survived'].sum() / base_set['Survived'].sum() * 100
```

# Самые популярные мужское и женское имена

```python
base_set['FullName'] = (base_set['Name'].str.split(',').str[1]
.str.split('.').str[1])

base_set['FirstName'] = np.where(
    base_set['FullName'].str.contains('(', regex=False),

    base_set['FullName'].str.split('(').str[1]
    .str.split(' ').str[0]
    .str.strip(),

    base_set['FullName'].str.split(' ').str[1]
    .str.strip()
)

top_female_name = base_set[base_set['Sex'] == 'female']['FirstName'].value_counts().idxmax()
top_male_name = base_set[base_set['Sex'] == 'male']['FirstName'].value_counts().idxmax()

print(f'The most popular female name was {top_female_name} and the most popular male name was {top_male_name}')
```

```python
top_names = (
    base_set
    .groupby('Sex')['FirstName']
    .agg(lambda x: x.value_counts().idxmax())
)

top_names
```

# Самые популярные мужское и женское имена по классам обслуживания

```python
top_names_pclass = (
    base_set
    .groupby(['Pclass', 'Sex'])['FirstName']
    .agg(lambda x: x.value_counts().idxmax())
)

top_names_pclass
```
